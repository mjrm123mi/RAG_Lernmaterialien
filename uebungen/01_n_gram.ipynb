```python
import random
from collections import defaultdict, Counter

# 1. Generate n-grams from text

# This is defined but never used!
def generate_ngrams(words, n):
    """Return list of n-grams (as tuples) from a list of words."""
    return [tuple(words[i:i + n]) for i in range(len(words) - n + 1)]

# 2. Create the model
def train_ngram_model(text, n):
    """
    Train an n-gram model from the given text.
    Returns a dictionary:
        { prefix (tuple): { next_word: probability, ... }, ... }
    """
    words = text.split()
    counts = defaultdict(Counter)

    # Count occurrences of (prefix → next_word)
    for i in range(len(words) - n + 1):
        prefix = tuple(words[i:i + n - 1])
        next_word = words[i + n - 1]
        counts[prefix][next_word] += 1

    # Convert counts to probabilities
    model = {}
    for prefix, next_words in counts.items():
        total = sum(next_words.values())
        model[prefix] = {word: count / total for word, count in next_words.items()}

    return model


# 3. Generate text from the model
def generate_text(model, n, num_words=50, seed=None):
    """
    Generate text using the trained n-gram model.
    """
    if not model:
        raise ValueError("Model is empty. Train it first!")

    # Choose random prefix or use seed
    if seed:
        prefix = tuple(seed.split())
    else:
        prefix = random.choice(list(model.keys()))

    output = list(prefix)

    for _ in range(num_words - len(prefix)):
        next_words = model.get(prefix)
        if not next_words:
            break

        words, probs = zip(*next_words.items())
        next_word = random.choices(words, weights=probs, k=1)[0]

        output.append(next_word)
        prefix = tuple(output[-(n - 1):])  # update prefix

    # Format output with line breaks
    lines = []
    for i in range(0, len(output), 10):
        lines.append(" ".join(output[i:i+10]))
    return "\n".join(lines)
```

```python
import requests

# Download text from Project Gutenberg
url = "https://www.gutenberg.org/cache/epub/1112/pg1112.txt"
text = requests.get(url).text

# Optional: clean up header/footer
start = text.find("The Tragedie of Romeo and Juliet")
end = text.find("*** END OF THE PROJECT GUTENBERG EBOOK ROMEO AND JULIET ***")
text = text[start:end]

print(text[:1000])  # preview first 1000 characters
```

```python
n = 2  # try 2 for bigrams, 3 for trigrams, etc.
model = train_ngram_model(text, n)
model
```

```python
generated = generate_text(model, n, num_words=50, seed="Juliet")
print(generated)
```
